```python
import sys
from pathlib import Path

sys.path.append(str(Path("../src").resolve()))

import joblib
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split

from data_loader import load_dataset
from preprocessing import split_features, build_preprocessor
from config import RANDOM_STATE, TEST_SIZE, MODEL_DIR
```

```python
X, y, df = load_dataset()

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE
)

print("Train:", X_train.shape)
print("Valid:", X_valid.shape)
```

```python
numeric_features, categorical_features = split_features(X_train)

print("Numeric:", len(numeric_features))
print("Categorical:", len(categorical_features))
```

```python
preprocessor = build_preprocessor(
    numeric_features,
    categorical_features
)
```

```python
alphas = [
    0.001,
    0.01,
    0.1,
    1,
    10,
    100,
    1000
]

results = []

for alpha in alphas:

    model = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("regressor", Ridge(alpha=alpha))
    ])

    model.fit(X_train, y_train)

    pred = model.predict(X_valid)

    mae = mean_absolute_error(y_valid, pred)

    results.append((alpha, mae))

    print(f"alpha={alpha:<6} MAE={mae:.2f}")
```

```python
results_df = pd.DataFrame(
    results,
    columns=["alpha", "MAE"]
)

results_df.sort_values("MAE")
```

```python
best_alpha = results_df.loc[
    results_df["MAE"].idxmin(),
    "alpha"
]

best_mae = results_df["MAE"].min()

print("Best alpha:", best_alpha)
print("Best MAE:", best_mae)
```

```python
best_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", Ridge(alpha=best_alpha))
])

best_model.fit(X_train, y_train)
```

```python
pred = best_model.predict(X_valid)

mae = mean_absolute_error(y_valid, pred)

print("Final Ridge MAE:", mae)
```

```python
joblib.dump(
    best_model,
    MODEL_DIR / "ridge_model.pkl"
)

print("Model saved.")
```

```python
X_numeric = X.select_dtypes(include=["number"])

print(X_numeric.shape)
```

```python
X_train_num, X_valid_num, y_train_num, y_valid_num = train_test_split(
    X_numeric,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE
)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

numeric_preprocessor = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
numeric_model = Pipeline(steps=[
    ("preprocessor", numeric_preprocessor),
    ("regressor", Ridge(alpha=0.1))
])
```

```python
numeric_model.fit(X_train_num, y_train_num)
```

```python
pred_num = numeric_model.predict(X_valid_num)

mae_num = mean_absolute_error(
    y_valid_num,
    pred_num
)

print("Numeric-only Ridge MAE:", mae_num)
```
